```python
import pandas as pd
import numpy as np
import torch
from sentence_transformers import SentenceTransformer
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print(device)  # should print "cuda"
embed_model = SentenceTransformer("all-MiniLM-L6-v2", device=device)
```

```python
import pandas as pd
import numpy as np

df_overview = pd.read_parquet("/kaggle/input/datasets/taarushgaba/overview-only-1-parquet")
df_overview["overview"] = df_overview["overview"].fillna("").astype(str).str.strip()

overview_embeddings = embed_model.encode(
    df_overview["overview"].tolist(),
    batch_size=256,
    show_progress_bar=True,
    convert_to_numpy=True
).astype(np.float16)

np.save("overview_embeddings.npy", overview_embeddings)
np.save("overview_ids.npy", df_overview["id"].to_numpy())
np.save("overview_content_types.npy", df_overview["content_type"].to_numpy())
```
